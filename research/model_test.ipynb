```python
ollama_llava = LLM(
            model="ollama/llava:7b",
            api_base="http://localhost:11434"
            )
```

```python
import base64
import numpy as np

def encode_image(image_path):
    """Converts an image to base64 format to send to Ollama."""
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")
```

```python
img_path = "test.png"
img_path = "image_1.png"
img_path = "image_2.png"


image_base64 = encode_image(img_path)
image_base64
```

```python
from crewai import Agent

image2text_agent = Agent(
    role="Image Analyst",
    goal="Analyze images and extract meaningful text or descriptions.",
    backstory="An AI-powered image analysis expert who can describe and interpret images.",
    llm=ollama_llava
)
```

```python
from crewai import Task

image2text_task = Task(
    description="Analyze an image and provide a textual description.",
    agent=image2text_agent,
    expected_output="A text description of the image."
)

```

```python
crew = Crew(
    agents=[image2text_agent],
    tasks=[image2text_task],
    verbose=True,
)
```

```python
result = crew.kickoff(inputs={'image_path': 'image_2.png'})
```

```python
import ollama

prompt = "Describe the content of this image."

response = ollama.chat(
    model="llava:7b",
    messages=[
        {
            "role": "user",
            "content": prompt,
            "images": [image_base64]  # Send the image as base64
        }
    ]
)
# Print the response
print(response['message']['content'])
```
